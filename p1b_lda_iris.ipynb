Train an LDA classifier to predict the Iris flower species from four measurements. Then test it and predict one new flower.

## 2. Create the required dataset

This is the relevant part of `make_data.py`, placed here so the notebook is self-contained.

```python
# Create the Iris dataset (built into scikit-learn) and save it as a CSV.
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True).frame.rename(columns={"target": "species"})
iris.to_csv("data/Iris.csv", index=False)
print("Created data/Iris.csv", iris.shape)
```

## 3. Practical implementation

The following cells perform the analysis and show the output.

```python
import pandas as pd
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix

df = pd.read_csv("data/Iris.csv")
X, y = df.drop(columns="species"), df["species"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

lda = LinearDiscriminantAnalysis().fit(X_train, y_train)
pred = lda.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print("Confusion matrix:\n", confusion_matrix(y_test, pred))

# classify one new flower (sepal l, sepal w, petal l, petal w)
new = pd.DataFrame([[5.1, 3.5, 1.4, 0.2]], columns=X.columns)
print("Predicted species (0=setosa,1=versicolor,2=virginica):", lda.predict(new)[0])
```
